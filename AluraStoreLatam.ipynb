# 2. Ventas por categoría

```python
tienda1.groupby("Categoría del Producto").agg(
    cantidad_productos = ("Producto", "count"),
    precio_total = ("Precio", "sum")
)
#agg()
#método de agregación que permite aplicar una o múltiples operaciones a columnas específicas de un DataFrame o Series
```

```python
def ventas_categorias(df):
    df_categoria = df.groupby("Categoría del Producto").agg(
        cantidad_productos = ("Producto", "count"),
        precio_total = ("Precio", "sum")
    )
    return df_categoria

v_categorias = []

for i,t in enumerate(tiendas):
    v_categorias.append(ventas_categorias(t))
    print("-------------------------------------------------------------------------------\n")
    print(f"Ventas por categorías de la tienda{i+1}\n")
    print("-------------------------------------------------------------------------------\n")
    print(ventas_categorias(t))
    print("-------------------------------------------------------------------------------\n")
```

# 3. Calificación promedio de la tienda


```python
def note_promedio(df):
    return df.mean()
cal_prom = []
for i, t in enumerate(tiendas):
    cal_prom.append(note_promedio(t["Calificación"]))
    print(f"La calificación promedio de la tienda {i+1} es: {note_promedio(t['Calificación'])}")
```

# 4. Productos más y menos vendidos

```python
producto_max = v_categorias[0]["cantidad_productos"].idxmax()
producto_max
```

```python
def producto_mas_menos_vendido(df):
    return df["cantidad_productos"].idxmax(), df["cantidad_productos"].max(), df["cantidad_productos"].idxmin(), df["cantidad_productos"].min()

grafico = {}

for i, t in enumerate(v_categorias):
    producto_max, cantidad_max, producto_min, cantidad_min = producto_mas_menos_vendido(t)
    grafico[f"tienda_{i+1}"] = {
        producto_max: cantidad_max,
        producto_min: cantidad_min
    }
    print("-------------------------------------------------------------------------------\n")
    print(f"El producto más vendido de la tienda {i+1} es {producto_max} con {cantidad_max} ventas")
    print(f"El producto menos vendido de la tienda {i+1} es {producto_min} con {cantidad_min} ventas")
    print("-------------------------------------------------------------------------------\n")
#
```

# 6. Anexo de análisis

```python
from matplotlib import pyplot as plt
import numpy as np
tienda1.head()
```

```python
def df_analisis(df):
    df_nuevo = df.drop(columns=["Fecha de Compra", "Vendedor", "Lugar de Compra", "Método de pago", "Cantidad de cuotas", "lat", "lon"])
    df_nuevo["Neto"] = df_nuevo["Precio"] - df_nuevo["Costo de envío"]
    df_nuevo_cat = df_nuevo.groupby("Categoría del Producto").agg(
        cantidad_productos = ("Producto", "count"),
        total_neto = ("Neto", "sum"),
        calificacion_promedio = ("Calificación", "mean")
    )
    return df_nuevo_cat
Ingresos_tiendas = []
for i, t in enumerate(tiendas):
    print(f"Análisis de la tienda {i+1}:\n")
    print(df_analisis(t))
    Ingresos_tiendas.append(df_analisis(t)["total_neto"].sum())
    print("-------------------------------------------------------------------------------\n")
```

```python
def plot_analisis_tiendas(tiendas):
    # Creamos una figura con dos subplots: uno para el gráfico de líneas y otro para el scatter
    plt.figure(figsize=(14, 6))

    #Gráfico de líneas
    plt.subplot(1, 2, 1)
    for idx, tienda in enumerate(tiendas):
        df_cat = df_analisis(tienda)

        plt.plot(df_cat.index, df_cat["total_neto"], marker="o", label=f"Tienda {idx+1}")

    plt.title("Ingresos Totales Netos por Categoría (por Tienda)")
    plt.xlabel("Categoría del Producto")
    plt.ylabel("Total Neto")
    plt.xticks(rotation=45, ha="right")
    plt.legend()
    plt.grid(True)

    # Gráfico de dispersión
    plt.subplot(1, 2, 2)
    for idx, tienda in enumerate(tiendas):
        df_cat = df_analisis(tienda)

        plt.scatter(df_cat["calificacion_promedio"], df_cat["total_neto"], s=50, alpha=0.8, label=f"Tienda {idx+1}")

        for i in range(len(df_cat)):
            plt.text(df_cat["calificacion_promedio"].iloc[i],
                     df_cat["total_neto"].iloc[i],
                     f"{df_cat.index[i]} (T{idx+1})",
                     fontsize=9, alpha=0.7)

    plt.title("Calificación Promedio vs Total Neto (por Tienda)")
    plt.xlabel("Calificación Promedio")
    plt.ylabel("Total Neto")
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.show()


plot_analisis_tiendas(tiendas)
```

```python
categorias = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
ventas = Ingresos_tiendas
satisfaccion = cal_prom

plt.figure(figsize=(14, 6))
plt.subplot(1, 2, 1)
# Crear el gráfico circular
plt.pie(ventas, labels=categorias, autopct='%1.1f%%', startangle=90, colors=plt.cm.Paired.colors)
plt.title('Distribución de ingresos netos por tienda')
plt.axis('equal')


plt.subplot(1, 2, 2)
plt.pie(satisfaccion, labels=categorias, autopct='%1.1f%%', startangle=90, colors=plt.cm.Paired.colors)
plt.title('Satisfacción de usuarios por tienda')
plt.axis('equal')

plt.tight_layout()
plt.show()
```

```python
from matplotlib import pyplot as plt
import seaborn as sns

def analisis_espacio(tiendas):

    plt.figure(figsize=(14, 6))
    for i, tienda in enumerate(tiendas):
        sns.kdeplot(
            data=tienda, x="lat", y="lon", fill=True,
            alpha=0.3
        )
    #Heatmap de densidad

    plt.title("Densidad espacial (Latitud vs Longitud)")
    plt.xlabel("Latitud")
    plt.ylabel("Longitud")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

analisis_espacio(tiendas)
```

```python
from matplotlib.lines import Line2D

def analisis_espacio2(tiendas):
    plt.figure(figsize=(14, 6))
    ax = plt.gca()

    colores = ['blue', 'orange', 'green', 'red']
    nombres = ['Tienda1', 'Tienda2', 'Tienda3', 'Tienda4']

    for i, tienda in enumerate(tiendas):
        # Línea de contorno con label para la leyenda
        sns.kdeplot(
            data=tienda, x="lat", y="lon",
            fill=False, color=colores[i], ax=ax
        )

    leyenda_manual = [Line2D([0], [0], color=color, lw=2, label=nombre)
                      for color, nombre in zip(colores, nombres)]
    ax.legend(handles=leyenda_manual, title="Tiendas")

    plt.title("Fronteras espaciales (Latitud vs Longitud)")
    plt.xlabel("Latitud")
    plt.ylabel("Longitud")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
analisis_espacio2(tiendas)
```
